```python
import torch
import torchaudio
from pydub import AudioSegment
from speechbrain.inference.speaker import SpeakerRecognition
import assemblyai as aai

device = "cuda" if torch.cuda.is_available() else "cpu"
print("Using device:", device)

verification = SpeakerRecognition.from_hparams(
    source="speechbrain/spkrec-ecapa-voxceleb",
    savedir="pretrained_ecapa",
    run_opts={"device": device},
)
```

```python
import assemblyai as aai
aai.settings.api_key = "...."

audio_file = "sample2.mp3"

config = aai.TranscriptionConfig(
  speaker_labels=True)

transcript = aai.Transcriber().transcribe(audio_file, config)
md_text = ""
for utterance in transcript.utterances:
  md_text += f"Speaker {utterance.speaker}: {utterance.text} \n"

```

```python
from collections import defaultdict

speaker_segments = defaultdict(list)

for utt in transcript.utterances:
    speaker_label = f"SPEAKER {utt.speaker}"
    segment = {
        "start_time": utt.start,
        "end_time": utt.end
    }
    speaker_segments[speaker_label].append(segment)

speaker_list = []
for speaker_label, segments in speaker_segments.items():
    speaker_list.append({
        "speaker_name": speaker_label,
        "time_duration": segments
    })

print(speaker_list)
```

```python
from pydub import AudioSegment

main_audio = AudioSegment.from_file(audio_file)

def slice_and_save(audio_seg: AudioSegment, speaker_name):
    outpath = f"{speaker_name}.wav"
    audio_seg.export(outpath, format="wav")
    return outpath

speaker_filepaths = {}
for spk_label, segments in speaker_segments.items():
    combined_audio = AudioSegment.empty()

    for seg in segments:
        start_ms = seg["start_time"]
        end_ms = seg["end_time"]
        combined_audio += main_audio[start_ms:end_ms]

    outpath = slice_and_save(combined_audio, spk_label)
    speaker_filepaths[spk_label] = outpath
```

```python
reference_speakers = {
    "Darshana": "/content/speakers/darshana.mp3",
    "Muskan": "/content/speakers/muskan.mp3",
    "Priyanka": "/content/speakers/priyanka.mp3"
}
```

```python
speaker_to_ref = {}
for diar_speaker, speaker_audio_path in speaker_filepaths.items():
    best_score = -1.0
    best_match = None

    for ref_name, ref_file in reference_speakers.items():
        score_tensor, is_same = verification.verify_files(speaker_audio_path, ref_file)
        score = score_tensor.item()
        if score > best_score:
            best_score = score
            best_match = ref_name
    if best_score>0.6:
      speaker_to_ref[diar_speaker] = best_match
    print(f"{diar_speaker} most likely matches {best_match} with a score={best_score:.4f}")
```

```python
final_transcript = ""

for utt in transcript.utterances:
    diar_label = f"SPEAKER {utt.speaker}"  # e.g., "SPEAKER_A"
    real_name = speaker_to_ref[diar_label]  # e.g., "Priyanka Shah"
    text = utt.text
    final_transcript += f"{real_name}: {text}\n"

print(final_transcript)
```
